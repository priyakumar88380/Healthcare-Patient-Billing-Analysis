```python
plt.figure(figsize=(10,5))

# Use this specific syntax to ensure Seaborn maps the data correctly
sns.histplot(x=insurance['bmi'], bins=30, kde=True, color='teal')

plt.title('Univariate Analysis: BMI Distribution (1338 Rows)')
plt.xlabel('BMI')
plt.ylabel('Frequency')
plt.show()
```

```python
plt.figure(figsize=(10,5))
sns.boxplot(x=insurance['charges'],color='salmon')
plt.title('Univariate Analysis:Charges Boxplot')
plt.xlabel('Insurance Charges')
plt.show()
```

```python
plt.figure(figsize=(10,6))
sns.boxplot(data=insurance,x='smoker',y='charges',palette='Set2')
plt.title('Bivariate Analysis: Charges by Smoker Status')
plt.xlabel('Is the person a Smoker?')
plt.ylabel('Insurance Charges')
plt.show()
```

```python
plt.figure(figsize=(10, 6))

# Plotting BMI against Charges
# 'hue' adds a third layer of info by coloring the dots by smoker status!
sns.scatterplot(data=insurance, x='bmi', y='charges', hue='smoker', palette='magma', alpha=0.7)

plt.title('Bivariate Analysis: BMI vs. Charges (Colored by Smoker)')
plt.xlabel('BMI')
plt.ylabel('Insurance Charges')
plt.show()
```

```python
plt.figure(figsize=(10, 8))

# We only calculate correlation for numerical columns
# .corr() finds the relationships, sns.heatmap draws it
correlation_matrix = insurance.select_dtypes(include=['number']).corr()

sns.heatmap(correlation_matrix, annot=True, cmap='coolwarm', fmt=".2f")

plt.title('Multivariate Analysis: Correlation Matrix')
plt.show()
```

```python
# Select only the columns that have numbers
numerical_columns = insurance.select_dtypes(include=['number']).columns

# Loop through each numerical column and create a chart
for col in numerical_columns:
    plt.figure(figsize=(8, 4))
    sns.histplot(insurance[col], kde=True, color='orchid')
    plt.title(f'Distribution of {col}')
    plt.show()
```
